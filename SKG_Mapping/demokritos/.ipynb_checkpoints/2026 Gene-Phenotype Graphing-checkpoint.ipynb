```ruby
require 'linkeddata'
require 'rdf/nquads'
require 'csv'

SIMPATHIC = RDF::Vocabulary.new('urn:simpathic:')
RDFS      = RDF::Vocabulary.new('http://www.w3.org/2000/01/rdf-schema#')

hpo_index = {}
CSV.foreach('./maps/cui_hpo_lookup.tsv', headers: true, col_sep: "\t") do |row|
  hpo_index[row['CUI']] = row
end

gene_index = {}
CSV.foreach('./maps/2026-gene-mappings.map', headers: true) do |row|
  gene_index[row['source']] = row
end

def parse_pmids(prov)
  return [] if prov.nil? || prov.strip.empty?
  prov.scan(/(\d+)_fullText_/).flatten
end

failures        = {}
graphing_errors = File.open('./graph/2026-gene-phenotype-errors.txt', 'w')

File.open('./graph/2026-gene-phenotype.nq.large', 'w') do |f|
  writer = RDF::Writer.for(:nquads).new(f)

  CSV.foreach('./raw-data/Gene-Phenotype triples.tsv', col_sep: "\t", headers: true, quote_char: "\x00") do |row|
    gene_cui  = row['Gene_id']
    pheno_cui = row['Phenotype_id']

    gene_row = gene_index[gene_cui]
    hpo_row  = hpo_index[pheno_cui]

    unless gene_row
      unless failures[gene_cui]
        failures[gene_cui] = true
        graphing_errors.puts "gene lookup failed\tUMLS:#{gene_cui}\t#{row['Gene']}"
      end
      next
    end
    unless hpo_row
      unless failures[pheno_cui]
        failures[pheno_cui] = true
        graphing_errors.puts "phenotype lookup failed\tUMLS:#{pheno_cui}\t#{row['Phenotype']}"
      end
      next
    end

    gene_uri    = RDF::URI.new(gene_row['geneid'])
    protein_uri = RDF::URI.new(gene_row['protein'])
    taxon       = RDF::URI.new(gene_row['taxon'])
    hp_id       = hpo_row['HPO'].sub(':', '_')
    hpo_uri     = RDF::URI.new("http://purl.obolibrary.org/obo/#{hp_id}")
    relation    = row['RELATION']
    pmids       = parse_pmids(row['PROVENANCE'])

    context_uri     = RDF::URI.new("urn:simpathic:context:#{gene_cui}_#{pheno_cui}")
    general_context = RDF::URI.new('urn:simpathic:context:all_metadata')

    graph = RDF::Repository.new

    graph << RDF::Statement.new(gene_uri, SIMPATHIC['associated-with'], hpo_uri,  graph_name: context_uri)
    graph << RDF::Statement.new(hpo_uri,  SIMPATHIC['associated-with'], gene_uri, graph_name: context_uri)

    gene_type      = RDF::URI.new('http://edamontology.org/data_2610')
    gene_core_type = RDF::URI.new('https://w3id.org/biolink/vocab/Gene')
    graph << RDF::Statement.new(gene_uri,       RDFS.label,               RDF::Literal.new(gene_row['recommended_full']), graph_name: context_uri)
    graph << RDF::Statement.new(gene_uri,       RDF.type,                 gene_type,                                      graph_name: context_uri)
    graph << RDF::Statement.new(gene_uri,       RDF.type,                 gene_core_type,                                 graph_name: context_uri)
    graph << RDF::Statement.new(gene_type,      RDFS.label,               RDF::Literal.new('NCBI Gene'),                  graph_name: context_uri)
    graph << RDF::Statement.new(gene_core_type, RDFS.label,               RDF::Literal.new('Gene'),                       graph_name: context_uri)
    graph << RDF::Statement.new(gene_uri,       SIMPATHIC['original-id'], RDF::Literal.new(gene_cui),                     graph_name: context_uri)
    graph << RDF::Statement.new(gene_uri,       SIMPATHIC['in-taxon'],    taxon,                                          graph_name: context_uri)

    protein_type      = RDF::URI.new('http://edamontology.org/data_2291')
    protein_core_type = RDF::URI.new('https://w3id.org/biolink/vocab/Protein')
    graph << RDF::Statement.new(gene_uri,          SIMPATHIC['associated-with'], protein_uri,                              graph_name: context_uri)
    graph << RDF::Statement.new(protein_uri,       SIMPATHIC['associated-with'], gene_uri,                                 graph_name: context_uri)
    graph << RDF::Statement.new(protein_uri,       RDFS.label,               RDF::Literal.new(gene_row['recommended_full']), graph_name: context_uri)
    graph << RDF::Statement.new(protein_uri,       RDF.type,                 protein_type,                                 graph_name: context_uri)
    graph << RDF::Statement.new(protein_uri,       RDF.type,                 protein_core_type,                            graph_name: context_uri)
    graph << RDF::Statement.new(protein_type,      RDFS.label,               RDF::Literal.new('UniProt'),                  graph_name: context_uri)
    graph << RDF::Statement.new(protein_core_type, RDFS.label,               RDF::Literal.new('Protein'),                  graph_name: context_uri)
    graph << RDF::Statement.new(protein_uri,       SIMPATHIC['original-id'], RDF::Literal.new(gene_cui),                   graph_name: context_uri)
    graph << RDF::Statement.new(protein_uri,       SIMPATHIC['in-taxon'],    taxon,                                        graph_name: context_uri)

    hpo_type      = RDF::URI.new('http://edamontology.org/data_3275')
    hpo_core_type = RDF::URI.new('https://w3id.org/biolink/vocab/Phenotype')
    graph << RDF::Statement.new(hpo_uri,       RDFS.label,               RDF::Literal.new(hpo_row['HPO_Name']), graph_name: context_uri)
    graph << RDF::Statement.new(hpo_uri,       RDF.type,                 hpo_type,                              graph_name: context_uri)
    graph << RDF::Statement.new(hpo_uri,       RDF.type,                 hpo_core_type,                         graph_name: context_uri)
    graph << RDF::Statement.new(hpo_type,      RDFS.label,               RDF::Literal.new('HPO Ontology Term'), graph_name: context_uri)
    graph << RDF::Statement.new(hpo_core_type, RDFS.label,               RDF::Literal.new('Phenotype'),         graph_name: context_uri)
    graph << RDF::Statement.new(hpo_uri,       SIMPATHIC['original-id'], RDF::Literal.new(pheno_cui),           graph_name: context_uri)

    graph << RDF::Statement.new(context_uri, SIMPATHIC['skg-source'],      RDF::Literal.new('DEMOKRITOS'), graph_name: general_context)
    graph << RDF::Statement.new(context_uri, SIMPATHIC['source-relation'], RDF::Literal.new(relation),     graph_name: general_context)
    pmids.each do |pmid|
      graph << RDF::Statement.new(context_uri, SIMPATHIC['evidence'],
                                  RDF::URI.new("https://pubmed.ncbi.nlm.nih.gov/#{pmid}/"), graph_name: general_context)
    end

    graph.each_statement { |stmt| writer << stmt }
  end

  writer.flush
end

graphing_errors.close
puts "Done — see ./graph/2026-gene-phenotype-errors.txt for failures"
```
