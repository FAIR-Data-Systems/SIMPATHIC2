```ruby
require 'csv'
require 'net/http'
require 'json'

# Collect unique (CUI, UMLS name) pairs from all six phenotype TSVs
tsv_files = Dir.glob('raw-data/*Phenotype*.tsv')
puts "Found files: #{tsv_files.map { |f| File.basename(f) }.join(', ')}"

cui_map = {}
tsv_files.each do |f|
  CSV.foreach(f, col_sep: "\t", headers: true) do |row|
    cui  = row['Phenotype_id']&.strip
    name = row['Phenotype']&.strip
    cui_map[cui] ||= name if cui
  end
end
puts "Unique CUIs: #{cui_map.size}"

# Map CUI -> HPO via Monarch Initiative entity API
def get_hpo(cui)
  uri = URI("https://api-v3.monarchinitiative.org/v3/api/entity/UMLS:#{cui}")
  res = Net::HTTP.get_response(uri)
  return ['', ''] unless res.code == '200'
  data = JSON.parse(res.body)
  (data['xrefs'] || []).each do |xref|
    id = xref.is_a?(Hash) ? xref['id'].to_s : xref.to_s
    return [id, data['name'].to_s] if id.start_with?('HP:')
  end
  ['', '']
rescue => e
  puts "  Error for #{cui}: #{e.message}"
  ['', '']
end

results = []
cui_map.each_with_index do |(cui, umls_name), i|
  puts "  #{i}/#{cui_map.size} ..." if (i % 50).zero?
  hpo_id, hpo_name = get_hpo(cui)
  results << { CUI: cui, UMLS_Name: umls_name, HPO: hpo_id, HPO_Name: hpo_name }
  sleep 0.1
end

mapped = results.count { |r| !r[:HPO].empty? }
puts "\nMapped: #{mapped} / #{results.size} (#{'%.1f' % (100.0 * mapped / results.size)}%)"

CSV.open('maps/cui_hpo_lookup.tsv', 'w', col_sep: "\t") do |csv|
  csv << %w[CUI UMLS_Name HPO HPO_Name]
  results.each { |r| csv << [r[:CUI], r[:UMLS_Name], r[:HPO], r[:HPO_Name]] }
end

puts "Written to maps/cui_hpo_lookup.tsv"
results.first(5)
```
